```python
from rdkit import Chem
```

```python
electrophiles = ['CCBr','CCCI', 'IC1CCCCC1', 'CC(=O)C=C', 'CCCC(Cl)=O', 'ClCl', 'I', 'COC(C)=O', 'C[C+](C)C', 'FB(F)F']
nucleophiles = ['CCO', 'CN', 'C#N-', 'CC[O-]', 'CC=C', 'O', 'N', 'CCS', 'c1ccccc1', 'C[Mg]Br', 'C=C[Mg]Br', 'C[Li]', '[BH4-]']
total = electrophiles + nucleophiles
```

```python
smiles = []
for mol in total:
    for mol2 in total:
            smiles.append(mol +"."+ mol2)
```

```python
with open('training_suggestions.txt', 'w') as file:
    for smile in smiles:
        file.write(smile + '\n')
```
